# Exploration du dataset IMDB-5000

Dataset principal qu'on va utiliser pour le projet (Option 2 de la roadmap).

```python
df_movies = pd.read_csv(DATA_PATH + 'movie_metadata.csv')

print("Dimensions :", df_movies.shape)
print("\nNombre de colonnes :", len(df_movies.columns))
print("\nColonnes disponibles :")
print(df_movies.columns.tolist())
```

```python
df_movies.head()
```

```python
df_movies.info()
```

```python
valeurs_manquantes = (df_movies.isnull().sum() / len(df_movies) * 100).round(2)
valeurs_manquantes_triees = valeurs_manquantes.sort_values(ascending=False)

print("Pourcentage de valeurs manquantes par colonne :")
print(valeurs_manquantes_triees)
```
